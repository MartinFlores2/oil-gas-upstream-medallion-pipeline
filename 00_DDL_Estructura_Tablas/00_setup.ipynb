```python
%sql
CREATE TABLE IF NOT EXISTS workspace.oil_gas_silver.telemetry_clean (
  well_id STRING NOT NULL,
  reading_ts TIMESTAMP NOT NULL,
  pressure_psi DOUBLE NOT NULL,
  flow_bpd DOUBLE NOT NULL,
  source_file STRING,
  silver_loaded_at TIMESTAMP
) USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS workspace.oil_gas_silver.telemetry_quarantine (
  well_id STRING,
  sensor_timestamp STRING,
  wellhead_pressure_raw STRING,
  flow_bpd_raw STRING,
  reject_reason STRING,
  source_file STRING,
  quarantined_at TIMESTAMP
) USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS workspace.oil_gas_gold.well_daily_summary (
  well_id STRING NOT NULL,
  reading_date DATE NOT NULL,
  readings_count BIGINT NOT NULL,
  avg_pressure_psi DOUBLE,
  min_pressure_psi DOUBLE,
  avg_flow_bpd DOUBLE,
  gold_loaded_at TIMESTAMP
) USING DELTA;
```
