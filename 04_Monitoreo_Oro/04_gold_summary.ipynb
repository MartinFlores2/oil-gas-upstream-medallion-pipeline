```python
%sql
INSERT OVERWRITE workspace.oil_gas_gold.well_daily_summary
SELECT well_id,
       CAST(reading_ts AS DATE) AS reading_date,
       COUNT(*) AS readings_count,
       AVG(pressure_psi) AS avg_pressure_psi,
       MIN(pressure_psi) AS min_pressure_psi,
       AVG(flow_bpd) AS avg_flow_bpd,
       current_timestamp() AS gold_loaded_at
FROM workspace.oil_gas_silver.telemetry_clean
GROUP BY well_id, CAST(reading_ts AS DATE);
```

```python
%sql
SELECT COUNT(*) AS rows_gold, SUM(readings_count) AS total_readings
FROM workspace.oil_gas_gold.well_daily_summary;
```

```python
%sql
SELECT well_id,
       ROUND(MIN(avg_pressure_psi), 1) AS lowest_daily_avg,
       ROUND(MAX(avg_pressure_psi), 1) AS highest_daily_avg,
       ROUND(AVG(avg_pressure_psi), 1) AS overall_avg
FROM workspace.oil_gas_gold.well_daily_summary
GROUP BY well_id
ORDER BY well_id;
```
