```python
%sql
USE control_produccion_pozos;
```

```python
%sql

-- ---------------------------------------------------------------------
-- VIEW 1: CAPA SEMÁNTICA - MONITOREO DE ALERTAS CRÍTICAS (ORO)
-- Esta vista entrega las paradas de planta activas directamente al panel
-- ---------------------------------------------------------------------

CREATE OR REPLACE VIEW control_produccion_pozos.vw_alertas_criticas_dashboard AS
SELECT 
    id_alerta,
    id_pozo,
    fecha_hora,
    presion_psi,
    caudal_bpd,
    tipo_anomalia,
    nivel_riesgo,
    requiere_parada_planta
FROM control_produccion_pozos.alertas_produccion_oro;
```

```python
%sql
-- ---------------------------------------------------------------------
-- VIEW 2: CAPA SEMÁNTICA - MÉTRICAS DE CALIDAD DE DATOS (SILVER/GOVERNANCE)
-- ¡Acá está tu criterio! Muestra qué porcentaje se limpió, organizó o descartó
-- ---------------------------------------------------------------------
CREATE OR REPLACE VIEW control_produccion_pozos.vw_calidad_datos_dashboard AS
SELECT 
    estado_sensor,
    COUNT(*) as cantidad_registros,
    ROUND((COUNT(*) / (SELECT COUNT(*) FROM control_produccion_pozos.pozos_plata)) * 100, 2) as porcentaje_del_total
FROM control_produccion_pozos.pozos_plata
GROUP BY estado_sensor;
```

```python
%sql
-- ---------------------------------------------------------------------
-- VIEW 3: CAPA SEMÁNTICA - TENDENCIA TEMPORAL DE PRESIÓN
-- Entrega la serie de tiempo limpia para el gráfico de líneas del operador
-- ---------------------------------------------------------------------
CREATE OR REPLACE VIEW control_produccion_pozos.vw_tendencia_presion_dashboard AS
SELECT 
    id_pozo,
    fecha_hora,
    presion_psi,
    caudal_bpd
FROM control_produccion_pozos.pozos_plata
WHERE estado_sensor = 'OK';
```
