```python
%sql
CREATE OR REPLACE VIEW workspace.oil_gas_gold.v_well_status_daily AS
SELECT well_id,
       reading_date,
       avg_pressure_psi,
       CASE WHEN avg_pressure_psi < 1000 THEN 'LOW_PRESSURE' ELSE 'NORMAL' END AS pressure_status
FROM workspace.oil_gas_gold.well_daily_summary;
```

```python
%sql
SELECT well_id, pressure_status, COUNT(*) AS days
FROM workspace.oil_gas_gold.v_well_status_daily
GROUP BY well_id, pressure_status
ORDER BY well_id, pressure_status;
```
