```python
%sql
USE control_produccion_pozos;

SELECT 
    estado_sensor,
    COUNT(*) as cantidad_registros,
    ROUND((COUNT(*) / (SELECT COUNT(*) FROM pozos_plata)) * 100, 2) as porcentaje_del_total
FROM pozos_plata
GROUP BY estado_sensor;
```

```python
%sql
SELECT 
    id_pozo,
    CAST(AVG(presion_psi) AS DECIMAL(10,2)) as presion_promedio,
    CAST(MIN(presion_psi) AS DECIMAL(10,2)) as presion_minima,
    CAST(MAX(presion_psi) AS DECIMAL(10,2)) as presion_maxima,
    CAST(AVG(caudal_bpd) AS DECIMAL(10,2)) as caudal_promedio,
    CAST(MIN(caudal_bpd) AS DECIMAL(10,2)) as caudal_minimo
FROM pozos_plata
WHERE estado_sensor = 'OK'
GROUP BY id_pozo;
```
