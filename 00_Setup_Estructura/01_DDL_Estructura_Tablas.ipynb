```python
%sql
-- 1. Creamos la Base de Datos para nuestro proyecto si no existe
CREATE DATABASE IF NOT EXISTS control_produccion_pozos;
USE control_produccion_pozos;
```

```python
%sql
-- 2. DDL - TABLA BRONCE (Datos Crudos / Raw)
-- Recibe todo como STRING porque el sensor puede mandar cualquier dato defectuosa
CREATE TABLE IF NOT EXISTS pozos_bronce (
    id_pozo STRING,
    timestamp_sensor STRING,
    presion_cabezal_raw STRING,
    caudal_bpd_raw STRING,
    fecha_ingesta_bronce TIMESTAMP
) USING DELTA;
```

```python
%sql
-- 3. DDL - TABLA PLATA (Datos Limpios / Cleaned)
-- Acá ya tenemos los tipos de datos correctos (Float, Timestamp) para poder operar
CREATE TABLE IF NOT EXISTS pozos_plata (
    id_pozo STRING,
    fecha_hora TIMESTAMP,
    presion_psi FLOAT,
    caudal_bpd FLOAT,
    estado_sensor STRING, -- 'OK' o 'ERROR_LECTURA'
    fecha_ingesta_plata TIMESTAMP
) USING DELTA;
```

```python
%sql
-- 4. DDL - TABLA ORO (Datos de Negocio / Alertas y Anomalías)
-- Esta es la tabla que lee el operario o el tablero de control
CREATE TABLE IF NOT EXISTS alertas_produccion_oro (
    id_alerta STRING,
    id_pozo STRING,
    fecha_hora TIMESTAMP,
    presion_psi FLOAT,
    caudal_bpd FLOAT,
    tipo_anomalia STRING, -- 'CAIDA_PRESION_CRITICA', 'CAIDA_CAUDAL', 'OPERACION_NORMAL'
    nivel_riesgo STRING,   -- 'ALTO', 'MEDIO', 'BAJO'
    requiere_parada_planta BOOLEAN
) USING DELTA;
```

```python
%sql
TRUNCATE TABLE control_produccion_pozos.pozos_bronce;
TRUNCATE TABLE control_produccion_pozos.pozos_plata;
```
